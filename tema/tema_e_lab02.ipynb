
# Analiza dataset-ului Titanic

În acest notebook analizăm dataset-ul Titanic folosind Python,
Pandas și Seaborn. Vom calcula statistici descriptive și vom
vizualiza relația dintre supraviețuire, sex și clasă.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
titanic = sns.load_dataset("titanic")
titanic.head()
```

## Statistici descriptive

Analizăm dimensiunea dataset-ului, tipurile de date și
statisticile descriptive.

```python
print(titanic.shape)
print(titanic.dtypes)
titanic.describe()
```

## Vizualizare date

Vizualizăm distribuția supraviețuirii în funcție de sex.

```python
sns.countplot(data=titanic, x="survived", hue="sex")
plt.title("Supraviețuire în funcție de sex")
plt.show()
```

## Concluzie

Analiza dataset-ului Titanic arată că femeile au avut o rată
de supraviețuire mai mare decât bărbații. De asemenea,
pasagerii din clasele superioare au avut șanse mai mari de
supraviețuire. Vizualizările confirmă diferențele dintre
grupuri și oferă o imagine clară asupra structurii datelor.
